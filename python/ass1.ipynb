```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x1 = np.linspace(-np.sqrt(2),np.sqrt(2),1000)
x2 = np.linspace(0,np.sqrt(2),1000)
#print(x1.shape)
trueFuncArray = []
trueX1Array = []
trueX2Array = []

for i in range(x1.shape[0]):
    for j in range(x2.shape[0]):
        if 2-x1[i]**2-x2[j]**2 >= 0:
            trueFuncArray.append(x1[i]+2*x2[j])
            trueX1Array.append(x1[i])
            trueX2Array.append(x2[j])
        
trueFuncArray= np.array(trueFuncArray)
trueX1Array=np.array(trueX1Array)
trueX2Array=np.array(trueX2Array)
```

```python
fGradient = np.array([1,2])
c1Gradient = np.array([2*np.sqrt(2),0])
c2Gradient = np.array([0,1])
```

```python
plt.scatter(trueX1Array,trueX2Array,c=trueFuncArray)
plt.ylabel("x2")
plt.xlabel("x1")

plt.quiver(-np.sqrt(2),0, fGradient[0], fGradient[1], color=['r'],scale=10)
plt.quiver(-np.sqrt(2),0, c1Gradient[0], c1Gradient[1], color=['y'],scale=10)
plt.quiver(-np.sqrt(2),0, c2Gradient[0], c2Gradient[1], color=['y'],scale=5)

plt.title("Feasible set")
plt.show()
```

```python
fGradient = np.array([2,1])
c1Gradient = np.array([-4*np.sqrt(2/5), -2*np.sqrt(2/5)])
c2Gradient = np.array([4*np.sqrt(2/5), 2*np.sqrt(2/5)])
```

```python
figure, axes = plt.subplots()
Drawing_colored_circle = plt.Circle(( 0 , 0 ), np.sqrt(2),fill=False)
 
axes.set_aspect(1)
axes.add_artist( Drawing_colored_circle )
axes.set_xlim(-3, 3)  
axes.set_ylim(-2, 2)  

plt.quiver(2*np.sqrt(2/5),np.sqrt(2/5), fGradient[0], fGradient[1], color=['r'],scale=10)
plt.quiver(2*np.sqrt(2/5),np.sqrt(2/5), c2Gradient[0], c2Gradient[1], color=['y'],scale=10)

plt.quiver(-2*np.sqrt(2/5),-np.sqrt(2/5), fGradient[0], fGradient[1], color=['r'],scale=10)
plt.quiver(-2*np.sqrt(2/5),-np.sqrt(2/5), c1Gradient[0], c1Gradient[1], color=['y'],scale=10)

axes.set_xlabel("x1")
axes.set_ylabel("x2")


plt.title('Feasible region')
plt.show()
```
